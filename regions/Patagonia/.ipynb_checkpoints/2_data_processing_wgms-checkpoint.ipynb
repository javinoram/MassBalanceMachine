# Procesamiento de Datos de MBM

Este notebook demuestra el flujo de trabajo de procesamiento de datos de **MassBalanceMachine** utilizando datos de glaciares peruanos.  
Te guía en la conversión de tus datos al formato **WGMS**, que se utilizará a lo largo de toda la cadena de procesamiento de MassBalanceMachine.  



```python
import pandas as pd
import geopandas as gpd

import sys
sys.path.append('../../../')
import massbalancemachine as mbm

cfg = mbm.Config()
```

<h2>1. Cargar el conjunto de datos de balance de masa superficial y recuperar el ID RGI por punto de medición</h2>
<p style='text-align: justify;'>
En este paso, definimos y cargamos nuestros datos glaciológicos de una región de interés. El conjunto de datos WGMS no incluye los IDs de RGI por defecto, por lo que necesitamos recuperarlos a partir de un archivo shapefile con los contornos de glaciares proporcionado por el Randolph Glacier Inventory (versión 6). Cada punto de medición (stake) se asocia con un ID RGI. El ID RGI es necesario para que el MassBalanceMachine pueda añadir características topográficas y meteorológicas adicionales durante la etapa de entrenamiento.
</p>
<p style='text-align: justify;'>
<b>Cómo recuperar los contornos de los glaciares:</b> Descarga los archivos shapefile para la región de interés desde este <a href='https://daacdata.apps.nsidc.org/pub/DATASETS/nsidc0770_rgi_v6/'>enlace</a>. Extrae los archivos y copia los archivos .shp, .prj, .shx y .dbf en el directorio correspondiente para poder usarlos con el Jupyter Notebook. Además, asegúrate de especificar correctamente el directorio y los archivos en la próxima celda de código.
</p>

<b>Nota:</b> Los registros de datos que contienen una FROM_DATE o TO_DATE inválida, donde el día o el mes están indicados como 99, serán eliminados del conjunto de datos.


```python
# Specify the filename of the input file with the raw data
target_data_fname = 'data/data-zona-norm.csv'
# Load the target data
data = pd.read_csv(target_data_fname, sep = ',')
display(data.head(30))
data['YEAR'] = pd.to_datetime(data["TO_DATE"], errors="coerce", dayfirst=True).dt.year
data["TO_DATE"] = pd.to_datetime(data["TO_DATE"], errors="coerce", dayfirst=True).dt.strftime("%Y%m%d")
data["FROM_DATE"] = pd.to_datetime(data["FROM_DATE"], errors="coerce", dayfirst=True).dt.strftime("%Y%m%d")
display(data.head(30))
```

<h3>1.1 Asociar las mediciones de los puntos con un ID RGI</h3>
<p style='text-align: justify;'>
Basado en la ubicación de la medición del punto indicada por POINT_LAT y POINT_LON, cada registro de datos se asocia con el ID RGI del glaciar donde se encuentra el punto de medición.
</p>

```python
# Especifique el nombre del archivo de forma del contorno de los glaciares obtenido de RGIv6
#glacier_outline_fname = 'data/rgi60/19rgi60/19_rgi60_AntarcticSubantarctic.shp'
glacier_outline_fname = 'rgi60/17_rgi60_SouthernAndes.shp'

# Cargar contornos de glaciares
glacier_outline = gpd.read_file(glacier_outline_fname)
display(glacier_outline.head(30))
```

```python
# Obtenga el ID de RGI para cada medición de estaca para la región de interés
data = mbm.data_processing.utils.get_rgi(data=data, glacier_outlines=glacier_outline)
display(data)
```

<p style='text-align: justify;'>
Luego, podemos crear un `Dataset` de MassBalanceMachine, utilizando el dataframe cargado con los datos de los puntos de medición junto con los IDs RGI correspondientes, de la siguiente manera:
</p>

```python
# Proporcione el nombre de la columna que contiene los ID de RGI para cada una de las estacas
#dataset = mbm.Dataset(cfg=cfg, data=data, region_name='CB', data_path='/home/lacrio/DATA/files/')
dataset = mbm.data_processing.Dataset(cfg=cfg, data=data, region_name='SouthernAndes', data_path='data/')
```

<h2>2. Obtener variables topográficas por medición de estaca</h2>
<p style='text-align: justify;'>
Una vez que hemos creado un <code>Dataset</code>, lo primero que podemos hacer es añadir datos topográficos a nuestro conjunto de datos. Esto se puede hacer automáticamente con MassBalanceMachine (que llama a OGGM) de la siguiente manera:
</p>

```python
# Especificar las variables topográficas de interés
# Consulte la documentación de OGGM para saber qué variables están disponibles: https://oggm.org/tutorials/stable/notebooks/10minutes/machine_learning.html ('topo', 'slope_factor', 'dis_from_border')
voi_topographical = ['aspect', 'slope']

# Recupere las variables topográficas para cada medición de estaca en función de la latitud y longitud de la estaca y agréguelas al conjunto de datos
dataset.get_topo_features(vois=voi_topographical)
```

```python
# Especifique los archivos de datos climáticos que se corresponderán con las coordenadas de los datos de estaca
era5_climate_data = './era5land/raw/era5_monthly_averaged_data.nc'
#geopotential_data = './data/era5land/raw/era5_geopotential_pressure_2.nc'

#era5_climate_data = '../../notebooks/example_data/iceland/climate/era5_monthly_averaged_data.nc'
geopotential_data = '../../notebooks/example_data/iceland/climate/era5_geopotential_pressure.nc'

# Haga coincidir las características climáticas, del archivo netCDF de ERA5Land, para cada conjunto de datos de medición de estaca
dataset.get_climate_features(climate_data=era5_climate_data, geopotential_data=geopotential_data) 
display(dataset)
```

<h2>4. Transformar los datos a una resolución temporal mensual</h2>
<p style='text-align: justify;'>
Finalmente, necesitamos transformar el conjunto de datos a una resolución mensual, teniendo en cuenta un período variable (FROM_DATE y TO_DATE) para los datos de SMB objetivo. Esto se hará para realizar predicciones de SMB en un paso temporal mensual, que luego se integrarán tanto en el tiempo como en el espacio para coincidir con las observaciones glaciológicas y geodésicas de SMB disponibles para el entrenamiento. Por favor, especifique las variables climáticas utilizadas en el paso anterior en la lista a continuación. Consulte la <a href='https://confluence.ecmwf.int/display/CKB/ERA5-Land%3A+data+documentation'>documentación</a> de todos los nombres abreviados de las variables climáticas.
</p>
<p style='text-align: justify;'>
Al tener en cuenta períodos variables, MassBalanceMachine utiliza el inicio de mes más cercano a FROM_DATE y TO_DATE e incluye todos los meses intermedios. Por ejemplo, FROM_DATE 19940917 y TO_DATE 19950520 incluirán todos los meses desde octubre de 1994 hasta mayo de 1995.
</p>

```python
# Especifique los nombres cortos de las variables climáticas disponibles en el conjunto de datos
vois_climate = ['t2m', 'tp', 'slhf', 'sshf', 'ssrd', 'fal', 'str']

# Para cada registro, conviértalo a una resolución de tiempo mensual
dataset.convert_to_monthly(vois_climate=vois_climate, vois_topographical=voi_topographical)
```

Finalmente, podemos echar un vistazo al conjunto de datos que se utilizará para el entrenamiento.
